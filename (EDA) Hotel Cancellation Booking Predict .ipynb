```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv('clean_data.csv')
df = df.drop('Unnamed: 0', axis = 1)
```

```python
df.head()
```

```python
plt.figure(figsize = (20,10))
sns.heatmap(df.corr(), annot = True)
```

```python
df.info()
```

```python
sns.jointplot(y = 'stays_in_week_nights', x = 'lead_time', data = df)
```

### Terdapat korelasi positif lemah , antara stays in week nights dengan lead time dapat dilihat dari jointplot di atas dan tabel korelasi heatmap. dimana korelasi tersebut bernilai 0.16

```python
plt.figure(figsize = (11,6))

n_cancel = df[df['is_canceled'] == 0]
sns.distplot(n_cancel['lead_time'], bins = 50)
plt.title('Distribusi lead time yang tidak melakukan cancel')
plt.xlabel('Lead Time')
#plt.savefig('fig-2')
```

```python
plt.figure(figsize = (11,6))

cancel = df[df['is_canceled'] == 1]
sns.distplot(cancel['lead_time'], bins = 50)
plt.title('Distribusi lead time yang melakukan cancel')
plt.xlabel('Lead Time')
#plt.savefig('fig-3')
```

### Berdasarkan kedua distribution plot lead time di atas antara yg tidak melakukan cancel dan melakukan cancel, dapat dilihat ada perbedaan antara keduanya. dimana untuk yg tidak melakukan cancel memiliki rata-rata lead time berkisar kurang lebih dari 0 - 30. Sedangkan yang melakukan cancel memiliki rata-rata lead time yang lebih tinggi berkisar kurang lebih antara 0 - 100, bahkan masih tingginya distribusi lead time diatas 100 dibandingkan dengan yg tidak melakukan cancel.

```python
plt.figure(figsize = (11, 6))

sns.barplot(x = 'market_segment', y = 'lead_time', hue = 'is_canceled', data = df)
plt.title('Lead time per market segment')
plt.legend()
#plt.savefig('fig-4')
```

### Rata_rata lead time tertinggi dimiliki oleh segment groups, dibandingkan dengan segment lainnya. Terutama yang melakukan cancel dibandingkan dengan yg tidak melakukan cancel.

```python
plt.figure(figsize=(11, 6))

sns.barplot(x = 'market_segment',
            y = 'adr',
            data = df
           )

plt.title('ADR per Market Segment')
plt.show()
#plt.savefig('fig-5.png')
```

### Rata-rata ADR tertinggi dimiliki oleh segment direct dan online TA, dibandingkan dengan Corporate. Karena mungkin corporate sudah memiliki kerjasama jangka panjang dengan pihak hotel yang memungkinkan tarif dapat lebih rendah.

```python
plt.figure(figsize=(11, 6))

sns.countplot(x = 'assigned_room_type', 
              hue = 'is_canceled', 
              data = df)

plt.title('Jumlah assign room type konsumen per Cancel dan Tidak Cancel ')
plt.show()
#plt.savefig('fig-6')
```

### Tipe kamar yang tertinggi dibooking ialah tipe kamar A, kemungkinan tipe kamar ini memang yang paling best seller atau memiliki kelas standar. Tipe tersebut jugalah yang memiliki jumlah cancel tertinggi.

```python
plt.figure(figsize=(11, 6))

sns.countplot(x = 'total_of_special_requests', 
              hue = 'is_canceled',
              data = df)

plt.title('Jumlah Konsumen Berdasarkan Total Special Request per Cancel dan Tidak Cancel ')
plt.show()
#plt.savefig('fig-7')
```

### Cancel tertinggi berada pada total special request 0, dari plot diatas dapat dilihat ada kecenderungan bahwa semakin rendah total request maka semakin tinggi jumlah cancel.

```python
plt.figure(figsize=(11, 6))

sns.countplot(x = 'deposit_type',
              hue = 'is_canceled',
              data = df)

plt.title('Jumlah Deposit Type per Cancel dan Tidak Cancel')
plt.show()
#plt.savefig('fig-8')
```

### Konsumen yang tidak cancel tertinggi berada pada tipe deposit No Deposit, begitu pula dengan yang melakukan cancel. Terlihat bahwa tipe refundable memiliki jumlah yang paling sedikit. kemungkinan terjadi karena para konsumen tidak terlalu menyukai pembayaran deposit diawal, jadi banyak yang memilih melakukan no deposit ketika melakukan booking.

```python
plt.figure(figsize=(11, 6))

sns.catplot(x = "deposit_type", 
            hue = "market_segment", 
            col = "is_canceled", 
            data = df, 
            kind = "count")

plt.suptitle(x = 0.5, y = 1.1, t = 'Jumlah Deposit Type Berdasarkan Cancel per Market segment', fontsize = 15)
plt.show()
#plt.savefig('fig-9')
```

### Untuk yang tidak melakukan cancel, jumlah terbanyak cenderung terdapat pada tipe deposit No Deposit, dengan jumlah tertinggi pada market segment Online TA (Tour Agent). sedangkan untuk yang melakukan cancel cenderung ada pada tipe deposit No Deposit & Non Refund dengan market segment Online TA juga memiliki jumlah tertinggi.

```python
is_canceled = df['is_canceled'].value_counts()

fig = px.pie(is_canceled, 
             values = is_canceled.values, 
             names = is_canceled.index, 
             title = 'Proporsi Antara Cancel dan Tidak Cancel', 
             template = 'seaborn')

fig.update_traces(rotation = -90, textinfo = 'value+percent')
fig.show()
#plt.savefig('fig-10')
```

### Berdasarkan pie chart di atas, proporsi antara yang tidak melakukan cancel dan cancel memiliki perbandingan kurang lebih 60:40 persen.

```python
canceled_df = df[df['is_canceled'] == 1]['reservation_status'].value_counts()

fig = px.pie(canceled_df, 
             values = canceled_df.values, 
             names = canceled_df.index, 
             title = 'Proporsi Konsumen yang Melakukan Cancel dibagi Antara Cancel & No-Show', 
             template = 'seaborn')

fig.update_traces(rotation = -90, textinfo = 'value+percent')
fig.show()
#plt.savefig('fig-11')
```

### Berdasarkan pie chart di atas, proporsi konsumen yang melakukan cancel jika dilihat dari reservation status nya terbanyak ialah yang melakukan cancel dengan pemberitahuan. Dibandingkan dengan yang tidak datang sama sekali tanpa pemberitahuan ingin melakukan cancel pada bookingnya. Dengan begitu kemungkinan dapat dengan mudah dilakukan sebuah evaluasi, misal dengan kuesioner untuk mengetahui alasan konsumen melakukan cancel pada bookingnya. Yang dapat dijadikan insight untuk kajian perusahaan lebih jauh, misalnya untuk mengetahui perilaku konsumen terhadap keputusan melakukan cancel booking.

```python
canceled_df = df[df['is_canceled'] == 1]['country'].value_counts()

fig = px.pie(canceled_df, 
             values = canceled_df.values, 
             names = canceled_df.index, 
             title = 'Proporsi Konsumen yang Melakukan Cancel dibagi Berdasarkan Negara Konsumen', 
             template = 'seaborn')

fig.update_traces(rotation = -90, textinfo = 'value+percent')
fig.show()
#plt.savefig('fig-12')
```

### Berdasarkan pie chart di atas, proporsi konsumen yang melakukan cancel tertinggi ada pada konsumen dengan kewarganegaraan Portugal. Hal ini terjadi karena memang dataset yang digunakan adalah dataset hotel yang berada di portugal.

```python
cust_type = df['customer_type'].value_counts()

fig = px.pie(cust_type, 
             values = cust_type.values, 
             names = cust_type.index, 
             title = 'Booking Berdasarkan Customer Type', 
             template = 'seaborn')

fig.update_traces(rotation = -90, textinfo = 'value+percent')
fig.show()
#plt.savefig('fig-13')
```

### Berdasarkan pie chart di atas, proporsi terbanyak untuk konsumen yang melakukan booking ialah customer type Transient. Dimana transient adalah konsumen yang melakukan stay dengan periode yang singkat. Dibandingkan dengan konsumen kontrak dan group.
